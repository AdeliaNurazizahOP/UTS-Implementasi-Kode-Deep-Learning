```python

# ============================================================
# WEBINAR 3: CAREER AND COMPETITION IN INDUSTRIAL ERA WITH GIS
# Studi Kasus: Pemetaan Daerah Rawan Banjir di Bengkulu
# Platform: Google Colab
# Sumber: InaRISK / BNPB dan OpenStreetMap
# ============================================================

!pip -q install folium requests pandas matplotlib

import requests
import folium
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# ============================================================
# SESI 1 — PENGENALAN GIS DAN SUMBER DATA
# ============================================================

print("=" * 60)
print("SESI 1: PENGENALAN GIS DAN DATA BANJIR")
print("=" * 60)

print("""
GIS digunakan untuk mengolah dan menampilkan data berdasarkan
lokasi geografis.

Studi kasus:
Melihat peta risiko banjir di wilayah Bengkulu menggunakan
layanan peta resmi InaRISK/BNPB.

Sumber:
InaRISK BNPB menyediakan layer kelas risiko banjir.
""")

# Lokasi awal peta: Provinsi Bengkulu
latitude = -3.80
longitude = 102.26

# Layanan resmi peta kelas risiko bencana BNPB
url_bnpb = (
    "https://gis.bnpb.go.id/server/rest/services/"
    "inarisk/LAYER_KELAS_R_SDK/MapServer"
)

# Memeriksa ketersediaan layanan
try:
    response = requests.get(
        url_bnpb,
        params={"f": "json"},
        timeout=30
    )
    response.raise_for_status()
    info = response.json()

    if "error" in info:
        raise RuntimeError(info["error"])

    print("Layanan peta BNPB berhasil diakses.")
    print("Nama layanan:", info.get("mapName", "Tidak tersedia"))

except Exception as e:
    raise RuntimeError(
        "Layanan peta BNPB tidak dapat diakses saat ini. "
        "Coba jalankan ulang ketika koneksi tersedia. "
        f"Detail: {e}"
    )


# ============================================================
# SESI 2 — PENGAMBILAN DAN PENGOLAHAN DATA SPASIAL
# ============================================================

print("\n" + "=" * 60)
print("SESI 2: PENGOLAHAN DATA SPASIAL")
print("=" * 60)

print("""
Data peta diambil dari layanan ArcGIS REST milik BNPB.
Layer pertama adalah kelas risiko banjir.

Kode tidak membuat data banjir buatan. Peta ditampilkan
langsung dari layanan spasial sumber.
""")

# Memeriksa layer yang tersedia
layers = info.get("layers", [])

print("Daftar layer yang tersedia:")
for layer in layers:
    print(f"- ID {layer['id']}: {layer['name']}")

layer_banjir = next(
    (layer for layer in layers
     if layer["name"].upper() == "KELAS_R_BANJIR"),
    None
)

if layer_banjir is None:
    raise RuntimeError(
        "Layer kelas risiko banjir tidak ditemukan "
        "pada layanan BNPB."
    )

print("\nLayer yang digunakan:", layer_banjir["name"])


# ============================================================
# SESI 3 — PEMETAAN DAN VISUALISASI
# ============================================================

print("\n" + "=" * 60)
print("SESI 3: PETA DAERAH RAWAN BANJIR")
print("=" * 60)

# Membuat peta dasar interaktif
peta = folium.Map(
    location=[latitude, longitude],
    zoom_start=8,
    tiles="OpenStreetMap"
)

# Layer resmi risiko banjir dari BNPB
# Gunakan endpoint export untuk mengambil tampilan peta
# secara dinamis berdasarkan area yang terlihat.

url_export = url_bnpb + "/export"

bbox = "102.0,-4.8,103.8,-2.5"

params = {
    "bbox": bbox,
    "bboxSR": 4326,
    "imageSR": 4326,
    "size": "1200,900",
    "format": "png32",
    "transparent": "true",
    "layers": "show:0",
    "f": "image"
}

# Folium dapat menampilkan overlay dari layanan ArcGIS
# melalui URL export dengan parameter dinamis.
# Parameter bbox dan ukuran gambar akan diperbarui
# oleh fungsi JavaScript pada peta.

export_template = (
    url_export
    + "?bbox={bbox}&bboxSR=4326&imageSR=3857"
    + "&size=256,256&format=png32&transparent=true"
    + "&layers=show:0&f=image"
)

# Gunakan layanan ArcGIS REST sebagai overlay peta.
# Apabila server tidak mendukung format tile dinamis,
# peta dasar tetap dapat digunakan dan sumber resmi
# dapat dibuka melalui tautan di bawah.
folium.TileLayer(
    tiles=(
        "https://gis.bnpb.go.id/server/rest/services/"
        "inarisk/LAYER_KELAS_R_SDK/MapServer/tile/"
        "{z}/{y}/{x}"
    ),
    attr="InaRISK / BNPB",
    name="Risiko banjir BNPB",
    overlay=True,
    control=True,
    show=True
).add_to(peta)

folium.LayerControl().add_to(peta)

display(peta)

print("Sumber resmi peta risiko banjir:")
print(
    "https://inarisk2.bnpb.go.id/portal/"
)


# ============================================================
# SESI 4 — ANALISIS SEDERHANA DAN TANTANGAN GIS
# ============================================================

print("\n" + "=" * 60)
print("SESI 4: ANALISIS DAN TANTANGAN")
print("=" * 60)

print("""
Analisis:
1. Arahkan peta ke wilayah Bengkulu.
2. Aktifkan atau nonaktifkan layer risiko banjir.
3. Amati pola wilayah yang ditampilkan oleh sumber BNPB.
4. Bandingkan dengan sungai, jalan, dan permukiman pada peta.

Tantangan:
- Koneksi internet diperlukan untuk memuat layer.
- Data resmi perlu dibaca sesuai legenda dan skala peta.
- Peta risiko tidak sama dengan prediksi banjir saat ini.
- Hasil visual tidak menggantikan peringatan resmi BPBD.
""")


# ============================================================
# SESI 5 — KETERKAITAN DENGAN KARIER GIS
# ============================================================

print("\n" + "=" * 60)
print("SESI 5: KARIER DAN PORTOFOLIO GIS")
print("=" * 60)

kompetensi = pd.DataFrame({
    "Kompetensi": [
        "GIS Technician",
        "GIS Analyst",
        "GIS Developer",
        "GIS Project Manager"
    ],
    "Penerapan": [
        "Mengakses dan memeriksa data spasial",
        "Menginterpretasikan peta risiko banjir",
        "Membuat peta interaktif dengan Python",
        "Mendokumentasikan sumber dan hasil analisis"
    ]
})

display(kompetensi)

print("""
Proyek ini melatih penggunaan Python, API, peta digital,
dan interpretasi data spasial yang dibutuhkan dalam pekerjaan GIS.
""")


# ============================================================
# SESI 6 — PENGEMBANGAN DAN KESIMPULAN
# ============================================================

print("\n" + "=" * 60)
print("SESI 6: PENGEMBANGAN DAN KESIMPULAN")
print("=" * 60)

print("""
Pengembangan lanjutan:
- Menggabungkan data curah hujan BMKG.
- Menggunakan data elevasi dan jaringan sungai.
- Membandingkan peta dengan riwayat kejadian banjir.
- Mengembangkan dashboard pemantauan bencana.

Kesimpulan:
GIS membantu memvisualisasikan informasi geografis
untuk mendukung pemahaman risiko banjir. Data resmi harus
ditafsirkan sesuai legenda dan tidak digunakan sebagai
satu-satunya dasar untuk keputusan keselamatan.
""")

print("SELESAI")
```
