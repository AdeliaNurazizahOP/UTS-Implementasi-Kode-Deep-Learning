```python

# ============================================================
# STUDI KASUS HPC DAN MACHINE LEARNING
# Estimasi Konsentrasi CO Berdasarkan Data Kualitas Udara
# Google Colab - Satu Notebook, 6 Sesi
# Dataset asli: UCI Machine Learning Repository
# ============================================================

import os
import time
import zipfile
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from io import BytesIO
from IPython.display import display, Markdown
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

pd.set_option("display.max_columns", None)
np.random.seed(42)


def materi(judul, isi):
    display(Markdown(f"## {judul}\n\n{isi}"))


# ============================================================
# SESI 1 - KONSEP HPC DAN DATASET
# ============================================================

materi(
    "SESI 1 — Konsep High Performance Computing (HPC)",
    """
    **Materi:**

    HPC adalah teknologi komputasi berkinerja tinggi untuk memproses
    perhitungan dan data dalam jumlah besar. HPC mendukung penelitian
    AI, simulasi ilmiah, analisis lingkungan, dan berbagai kebutuhan
    komputasi kompleks.

    Kinerja komputasi dapat dinyatakan dalam FLOPS
    (*Floating Point Operations per Second*).

    **Studi kasus:** Mengolah data kualitas udara untuk memperkirakan
    konsentrasi karbon monoksida (CO).

    **Catatan:** Praktik ini menggunakan Google Colab untuk memperagakan
    pengolahan data dan komputasi paralel sederhana, bukan menjalankan
    pekerjaan pada superkomputer sungguhan.
    """
)

print("Mengunduh dataset asli dari UCI Machine Learning Repository...")

url = (
    "https://archive.ics.uci.edu/"
    "ml/machine-learning-databases/00360/AirQualityUCI.zip"
)

response = requests.get(url, timeout=60)
response.raise_for_status()

with zipfile.ZipFile(BytesIO(response.content)) as arsip:
    nama_csv = next(
        nama for nama in arsip.namelist()
        if nama.lower().endswith("airqualityuci.csv")
    )
    with arsip.open(nama_csv) as file:
        df = pd.read_csv(
            file,
            sep=";",
            decimal=",",
            encoding="utf-8-sig",
            skipfooter=1,
            engine="python"
        )

# Hilangkan kolom kosong bawaan file
df = df.dropna(axis=1, how="all")
df.columns = df.columns.str.strip()

# Hapus baris yang seluruh isinya kosong
df = df.dropna(axis=0, how="all").reset_index(drop=True)

print("\nUkuran dataset:", df.shape)
print("\nContoh data asli:")
display(df.head())

print("\nDaftar kolom:")
print(df.columns.tolist())


# ============================================================
# SESI 2 - MASALAH LOKAL DAN PENGOLAHAN DATA
# ============================================================

materi(
    "SESI 2 — Pemanfaatan HPC untuk Permasalahan Lingkungan",
    """
    **Materi:**

    Analisis kualitas udara dapat membantu penelitian lingkungan.
    Sensor mengukur respons terhadap berbagai gas serta kondisi
    seperti suhu dan kelembapan.

    Dataset ini berasal dari pengukuran sensor kualitas udara
    di sebuah wilayah perkotaan di Italia pada 2004–2005.
    Dataset bukan pengukuran kondisi udara terkini di Indonesia.

    **Tujuan praktik:** Membersihkan nilai sensor yang tidak valid
    sebelum data digunakan untuk melatih model.
    """
)

# Nilai -200 menandakan data hilang pada dataset ini
df = df.replace(-200, np.nan)

# Kolom yang digunakan sebagai target
target = "CO(GT)"

# Fitur sensor dan kondisi lingkungan.
# Kolom gas referensi lain tidak digunakan agar target tidak
# diperkirakan langsung dari hasil pengukuran referensi gas lain.
fitur = [
    "PT08.S1(CO)",
    "PT08.S2(NMHC)",
    "PT08.S3(NOx)",
    "PT08.S4(NO2)",
    "PT08.S5(O3)",
    "T",
    "RH",
    "AH"
]

# Pastikan nama kolom tersedia
kolom_wajib = fitur + [target]
kolom_tidak_ada = [
    kolom for kolom in kolom_wajib
    if kolom not in df.columns
]

if kolom_tidak_ada:
    raise ValueError(
        "Kolom tidak ditemukan pada dataset: "
        + str(kolom_tidak_ada)
    )

# Ubah kolom menjadi numerik
for kolom in kolom_wajib:
    df[kolom] = pd.to_numeric(df[kolom], errors="coerce")

# Target harus tersedia; fitur boleh memiliki nilai hilang
df = df.dropna(subset=[target]).reset_index(drop=True)

X = df[fitur].copy()
y = df[target].copy()

print("Jumlah baris setelah target kosong dihapus:", len(df))
print("\nJumlah nilai hilang pada fitur:")
display(X.isna().sum().to_frame("Nilai Hilang"))

# Pembagian kronologis 80:20.
# Urutan data dipertahankan agar baris yang lebih akhir
# menjadi data pengujian.
batas = int(len(df) * 0.8)

X_train = X.iloc[:batas].copy()
X_test = X.iloc[batas:].copy()
y_train = y.iloc[:batas].copy()
y_test = y.iloc[batas:].copy()

# Imputasi hanya belajar dari data pelatihan
imputer = SimpleImputer(strategy="median")
X_train_bersih = imputer.fit_transform(X_train)
X_test_bersih = imputer.transform(X_test)

print("\nJumlah data pelatihan:", len(X_train))
print("Jumlah data pengujian:", len(X_test))
print("Jumlah fitur:", len(fitur))

print("\nContoh data yang sudah dibersihkan:")
display(pd.DataFrame(
    X_train_bersih,
    columns=fitur
).head())


# ============================================================
# SESI 3 - CARA KERJA HPC DAN MACHINE LEARNING
# ============================================================

materi(
    "SESI 3 — Pelatihan Model Machine Learning",
    """
    **Materi:**

    Alur kerja komputasi meliputi persiapan data, penentuan sumber
    daya, pelaksanaan program, dan analisis hasil.

    Random Forest Regression menggabungkan prediksi dari banyak
    pohon keputusan untuk memperkirakan nilai numerik.

    Pada studi kasus ini, model memperkirakan konsentrasi CO
    berdasarkan respons sensor dan kondisi lingkungan.
    """
)

print("Melatih Random Forest Regression...")

# Satu proses pekerja
waktu_mulai = time.perf_counter()

model_serial = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=1
)

model_serial.fit(X_train_bersih, y_train)

waktu_serial = time.perf_counter() - waktu_mulai

prediksi = model_serial.predict(X_test_bersih)

mae = mean_absolute_error(y_test, prediksi)
rmse = np.sqrt(mean_squared_error(y_test, prediksi))
r2 = r2_score(y_test, prediksi)

print("\nHASIL EVALUASI MODEL")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
print(f"Waktu pelatihan satu proses: {waktu_serial:.2f} detik")

print("""
Interpretasi:
- MAE: rata-rata besar kesalahan absolut prediksi.
- RMSE: ukuran kesalahan yang memberi penalti lebih besar
  pada kesalahan yang besar.
- R²: ukuran kemampuan model menjelaskan variasi target.
  Nilai negatif berarti model lebih buruk daripada baseline
  yang selalu memprediksi rata-rata target pada data uji.
""")

hasil = pd.DataFrame({
    "CO Aktual (mg/m³)": y_test.to_numpy(),
    "CO Prediksi (mg/m³)": prediksi
})

print("Contoh hasil pada data pengujian:")
display(hasil.head(10))


# ============================================================
# SESI 4 - KOMPETENSI DAN KOMPUTASI PARALEL
# ============================================================

materi(
    "SESI 4 — Komputasi Paralel dan Kompetensi",
    """
    **Materi:**

    Komputasi paralel membagi pekerjaan agar beberapa proses
    dapat dikerjakan secara bersamaan. Dalam Random Forest,
    pohon keputusan dapat dilatih menggunakan beberapa pekerja.

    Praktik ini membandingkan konfigurasi satu pekerja dan
    konfigurasi beberapa pekerja. Jumlah pekerja yang tersedia
    bergantung pada sumber daya runtime Google Colab.
    """
)

jumlah_pekerja = os.cpu_count() or 2
print("Jumlah CPU logis yang terdeteksi:", jumlah_pekerja)

waktu_mulai = time.perf_counter()

model_paralel = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

model_paralel.fit(X_train_bersih, y_train)

waktu_paralel = time.perf_counter() - waktu_mulai

prediksi_paralel = model_paralel.predict(X_test_bersih)

hasil_perbandingan = pd.DataFrame({
    "Konfigurasi": [
        "Satu pekerja",
        "Paralel (n_jobs=-1)"
    ],
    "Waktu pelatihan (detik)": [
        waktu_serial,
        waktu_paralel
    ],
    "MAE": [
        mae,
        mean_absolute_error(y_test, prediksi_paralel)
    ],
    "RMSE": [
        rmse,
        np.sqrt(mean_squared_error(y_test, prediksi_paralel))
    ],
    "R²": [
        r2,
        r2_score(y_test, prediksi_paralel)
    ]
})

print("\nPERBANDINGAN KONFIGURASI")
display(hasil_perbandingan.round(4))

if waktu_paralel < waktu_serial:
    print("Konfigurasi paralel lebih cepat pada pengujian ini.")
elif waktu_paralel > waktu_serial:
    print("Konfigurasi satu pekerja lebih cepat pada pengujian ini.")
else:
    print("Kedua konfigurasi memiliki waktu yang hampir sama.")

print(
    "\nCatatan: hasil waktu bergantung pada kondisi runtime, "
    "jumlah CPU, dan overhead pemrosesan. Perbandingan ini "
    "bukan tolok ukur kinerja superkomputer."
)


# ============================================================
# SESI 5 - TANTANGAN IMPLEMENTASI AIoT
# ============================================================

materi(
    "SESI 5 — Tantangan AIoT dan Kualitas Data",
    """
    **Materi:**

    Sistem AIoT menggabungkan sensor, jaringan, penyimpanan data,
    dan model AI. Tantangannya meliputi data hilang, kalibrasi
    sensor, kestabilan koneksi, keamanan, serta keterbatasan
    komputasi dan energi pada perangkat lokal.

    Pada sesi ini, periksa kelengkapan data dan pengaruhnya
    terhadap proses analisis.
    """
)

print("Ringkasan kualitas data sebelum imputasi:")
ringkasan = pd.DataFrame({
    "Nilai Hilang": X.isna().sum(),
    "Persentase Hilang (%)": (
        X.isna().mean() * 100
    ).round(2)
})
display(ringkasan)

print("\nStatistik data pelatihan setelah imputasi:")
display(pd.DataFrame(
    X_train_bersih,
    columns=fitur
).describe().round(2))

# Grafik perbandingan hasil prediksi dengan nilai aktual
jumlah_tampil = min(100, len(y_test))

plt.figure(figsize=(12, 5))
plt.plot(
    np.arange(jumlah_tampil),
    y_test.iloc[:jumlah_tampil].to_numpy(),
    label="CO aktual"
)
plt.plot(
    np.arange(jumlah_tampil),
    prediksi[:jumlah_tampil],
    label="CO prediksi"
)
plt.title("Perbandingan CO Aktual dan Prediksi pada Data Uji")
plt.xlabel("Urutan observasi pada data uji")
plt.ylabel("Konsentrasi CO (mg/m³)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# ============================================================
# SESI 6 - EVALUASI DAN INSIGHT PENELITIAN
# ============================================================

materi(
    "SESI 6 — Evaluasi dan Insight Penelitian",
    """
    **Materi:**

    Penelitian berbasis AI harus mengevaluasi kualitas data,
    pemilihan metode, hasil pengujian, dan keterbatasan model.
    Hasil eksperimen harus dilaporkan berdasarkan nilai aktual,
    bukan asumsi atau angka yang dibuat sebelumnya.

    Studi kasus ini merupakan latihan estimasi konsentrasi CO,
    bukan sistem peringatan resmi atau pengukuran kondisi udara
    secara langsung.
    """
)

print("KESIMPULAN EKSPERIMEN")
print(f"- Total observasi yang digunakan: {len(df)}")
print(f"- Data pelatihan: {len(X_train)}")
print(f"- Data pengujian: {len(X_test)}")
print(f"- Fitur yang digunakan: {len(fitur)}")
print(f"- MAE model: {mae:.4f} mg/m³")
print(f"- RMSE model: {rmse:.4f} mg/m³")
print(f"- R² model: {r2:.4f}")
print(f"- Waktu pelatihan satu pekerja: {waktu_serial:.2f} detik")
print(f"- Waktu pelatihan paralel: {waktu_paralel:.2f} detik")

print("""
INSIGHT:
1. Dataset nyata perlu diperiksa dan dibersihkan sebelum
   digunakan dalam Machine Learning.
2. Evaluasi model diperlukan untuk mengetahui kesalahan prediksi.
3. Pemrosesan paralel tidak selalu lebih cepat untuk setiap
   ukuran data dan kondisi komputer.
4. Hasil dari data historis luar negeri tidak otomatis berlaku
   untuk kondisi kualitas udara di Indonesia.
5. Penelitian lanjutan dapat menggunakan data lokal yang
   terukur, menguji metode lain, dan mengevaluasi model pada
   periode waktu yang berbeda.
""")

print("Seluruh sesi selesai dijalankan.")
print("Sumber dataset: UCI Machine Learning Repository")
print("https://archive.ics.uci.edu/dataset/360/air+quality")
```
